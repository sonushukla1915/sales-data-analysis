```python
import pandas as pd
import matplotlib.pyplot as plt

# Load CSV
df = pd.read_csv(
    r"C:\Users\UNIQUE CHOICE\OneDrive\Desktop\Sales-Data-Analysis\sales_data.csv"
)

# Show column names
print("Columns:")
print(df.columns.tolist())

# Basic information
print("\nData Info:")
print(df.info())

# Missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Duplicate rows
print("\nDuplicate Rows:")
print(df.duplicated().sum())

# Create Revenue column
df["revenue"] = df["quantity"] * df["price"]

# Total Revenue
print("\nTotal Revenue:")
print(df["revenue"].sum())

# Total Quantity
print("\nTotal Quantity Sold:")
print(df["quantity"].sum())

# Average Order Value
print("\nAverage Order Value:")
print(df["revenue"].mean())

# Product-wise Revenue
print("\nProduct-wise Revenue:")
print(
    df.groupby("product")["revenue"]
      .sum()
      .sort_values(ascending=False)
)

# City-wise Revenue
print("\nCity-wise Revenue:")
print(
    df.groupby("city")["revenue"]
      .sum()
      .sort_values(ascending=False)
)

# Category-wise Revenue
print("\nCategory-wise Revenue:")
print(
    df.groupby("category")["revenue"]
      .sum()
      .sort_values(ascending=False)
)

# Customer-wise Revenue
print("\nCustomer-wise Revenue:")
print(
    df.groupby("customer_name")["revenue"]
      .sum()
      .sort_values(ascending=False)
)
```

```python
import matplotlib.pyplot as plt

# 1. Product-wise Revenue
product_sales = df.groupby("product")["revenue"].sum().sort_values(ascending=False)

plt.figure(figsize=(8, 5))
product_sales.plot(kind="bar")
plt.title("Product-wise Revenue")
plt.xlabel("Product")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# 2. City-wise Revenue
city_sales = df.groupby("city")["revenue"].sum().sort_values(ascending=False)

plt.figure(figsize=(8, 5))
city_sales.plot(kind="bar")
plt.title("City-wise Revenue")
plt.xlabel("City")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# 3. Category-wise Revenue
category_sales = df.groupby("category")["revenue"].sum().sort_values(ascending=False)

plt.figure(figsize=(8, 5))
category_sales.plot(kind="bar")
plt.title("Category-wise Revenue")
plt.xlabel("Category")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# ============================================
# SALES ANALYTICS - VISUAL DASHBOARD
# ============================================

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# -----------------------------
# 1. Prepare Data
# -----------------------------

df["revenue"] = df["quantity"] * df["price"]

# -----------------------------
# 2. KPI SUMMARY
# -----------------------------

total_revenue = df["revenue"].sum()
total_quantity = df["quantity"].sum()
total_orders = len(df)
average_order_value = df["revenue"].mean()

print("=" * 50)
print("          SALES ANALYTICS DASHBOARD")
print("=" * 50)

print(f"💰 Total Revenue       : ₹{total_revenue:,.2f}")
print(f"📦 Total Quantity      : {total_quantity}")
print(f"🧾 Total Orders        : {total_orders}")
print(f"📊 Average Order Value : ₹{average_order_value:,.2f}")

print("=" * 50)


# -----------------------------
# 3. PRODUCT REVENUE
# -----------------------------

product_sales = (
    df.groupby("product")["revenue"]
      .sum()
      .sort_values(ascending=False)
)

plt.figure(figsize=(10, 5))

sns.barplot(
    x=product_sales.index,
    y=product_sales.values
)

plt.title("Product-wise Revenue", fontsize=16, fontweight="bold")
plt.xlabel("Product")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# -----------------------------
# 4. CITY REVENUE
# -----------------------------

city_sales = (
    df.groupby("city")["revenue"]
      .sum()
      .sort_values(ascending=False)
)

plt.figure(figsize=(10, 5))

sns.barplot(
    x=city_sales.index,
    y=city_sales.values
)

plt.title("City-wise Revenue", fontsize=16, fontweight="bold")
plt.xlabel("City")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# -----------------------------
# 5. CATEGORY REVENUE
# -----------------------------

category_sales = (
    df.groupby("category")["revenue"]
      .sum()
      .sort_values(ascending=False)
)

plt.figure(figsize=(8, 5))

plt.pie(
    category_sales.values,
    labels=category_sales.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Revenue Distribution by Category",
          fontsize=16,
          fontweight="bold")

plt.tight_layout()
plt.show()


# -----------------------------
# 6. SALES HEATMAP
# -----------------------------

heatmap_data = pd.pivot_table(
    df,
    values="revenue",
    index="city",
    columns="category",
    aggfunc="sum",
    fill_value=0
)

plt.figure(figsize=(10, 6))

sns.heatmap(
    heatmap_data,
    annot=True,
    fmt=".0f",
    linewidths=0.5
)

plt.title(
    "Sales Heatmap — City vs Category",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Category")
plt.ylabel("City")

plt.tight_layout()
plt.show()


# -----------------------------
# 7. QUANTITY HEATMAP
# -----------------------------

quantity_heatmap = pd.pivot_table(
    df,
    values="quantity",
    index="city",
    columns="category",
    aggfunc="sum",
    fill_value=0
)

plt.figure(figsize=(10, 6))

sns.heatmap(
    quantity_heatmap,
    annot=True,
    fmt=".0f",
    linewidths=0.5
)

plt.title(
    "Quantity Heatmap — City vs Category",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Category")
plt.ylabel("City")

plt.tight_layout()
plt.show()


# -----------------------------
# 8. TOP PRODUCTS
# -----------------------------

top_products = (
    df.groupby("product")["revenue"]
      .sum()
      .sort_values(ascending=False)
      .head(5)
)

plt.figure(figsize=(9, 5))

sns.barplot(
    x=top_products.values,
    y=top_products.index
)

plt.title(
    "Top 5 Products by Revenue",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Revenue")
plt.ylabel("Product")

plt.tight_layout()
plt.show()


# -----------------------------
# 9. CUSTOMER REVENUE
# -----------------------------

customer_sales = (
    df.groupby("customer_name")["revenue"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=customer_sales.values,
    y=customer_sales.index
)

plt.title(
    "Top Customers by Revenue",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Revenue")
plt.ylabel("Customer")

plt.tight_layout()
plt.show()


# -----------------------------
# 10. FINAL INSIGHTS
# -----------------------------

print("\n" + "=" * 50)
print("              KEY INSIGHTS")
print("=" * 50)

print(
    f"🏆 Top Product: "
    f"{product_sales.idxmax()} "
    f"(₹{product_sales.max():,.2f})"
)

print(
    f"🏙️ Top City: "
    f"{city_sales.idxmax()} "
    f"(₹{city_sales.max():,.2f})"
)

print(
    f"📂 Top Category: "
    f"{category_sales.idxmax()} "
    f"(₹{category_sales.max():,.2f})"
)

print(
    f"👤 Top Customer: "
    f"{df.groupby('customer_name')['revenue'].sum().idxmax()}"
)

print("=" * 50)
```
